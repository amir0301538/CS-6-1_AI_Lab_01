We are implementing list

```python
my_list = [1, 2, 3, 4, 5]
print(my_list)
```

append() Add an element at the end of the list

```python
my_list.append(6)
print(my_list)
```

clear() Removes all the elements from the list

```python
my_list.clear()
print(my_list)
```

copy() Returns a copy of the list

```python
my_list = [1, 2, 3, 4, 5]
list_copy = my_list.copy()
print(my_list)
```

count() Returns the number of times a specified value appears in the list

```python
count = my_list.count(3)
print(my_list)
```

extend() Adds elements from another list (or iterable) to the end of the current list

```python
my_list.extend([6, 7])
print(my_list)
```

index() Returns the index of the first occurrence of the specified value

```python
index = my_list.index(4)
```

insert() Adds an element at a specified position in the list

```python
my_list.insert(2, 9)
print(my_list)
```

pop() Removes and returns the element at the specified index

```python
popped_element = my_list.pop(2)
print(my_list)
```

remove() Removes the first occurrence of the specified value

```python
my_list.remove(3)
print(my_list)
```

reverse() Reverses the order of the *list*

```python
my_list.reverse()
print(my_list)
```

sort() Sorts the list in ascending order

```python
my_list.sort()
print(my_list)
```
