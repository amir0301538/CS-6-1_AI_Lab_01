```python
# Required Libraries
import numpy as np
from sklearn.model_selection import train_test_split as split_data
from sklearn.neighbors import KNeighborsClassifier as KNN
from sklearn.metrics import accuracy_score as acc_score
from sklearn.datasets import load_iris, make_classification

# =========================
# FUNCTION FOR TRAINING + TESTING
# =========================
def run_knn(X, y, dataset_name):

    # Split data
    X_train, X_test, y_train, y_test = split_data(
        X, y, test_size=0.3, random_state=42
    )

    print("\nResults for:", dataset_name)
    print("--------------------------")

    k_values = [1, 3, 5, 7]

    for value in k_values:

        classifier = KNN(n_neighbors=value)
        classifier.fit(X_train, y_train)

        predictions = classifier.predict(X_test)

        result = acc_score(y_test, predictions)

        print(f"K value = {value} --> Accuracy = {result:.4f}")


# =========================
# DATASET 1 (IRIS)
# =========================
iris_data = load_iris()

features_1 = iris_data.data
labels_1 = iris_data.target

run_knn(features_1, labels_1, "IRIS Dataset")


# =========================
# DATASET 2 (CUSTOM GENERATED)
# =========================
features_2, labels_2 = make_classification(
    n_samples=320,       # changed value
    n_features=6,        # changed value
    n_classes=3,
    n_clusters_per_class=1,
    random_state=42
)

run_knn(features_2, labels_2, "Synthetic Dataset")
```
